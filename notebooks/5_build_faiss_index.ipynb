```python
import numpy as np
import faiss
```

```python
PROJECT_DIR = '/content/drive/MyDrive/product_search_engine'
```

Load the embeddings

```python
# Load the embedding saved earlier
text_embeddings = np.load(f"{PROJECT_DIR}/embeddings/text_embeddings.npy")
image_embeddings = np.load(f"{PROJECT_DIR}/embeddings/image_embeddings.npy")
```

Print the shape

```python
print(f"Text Embeddings shape : {text_embeddings.shape}")
print(f"Image Embeddings shape : {image_embeddings.shape}")
```

## Build the FAISS index

```python
# Get the dimension of embedding vectors
dimension = text_embeddings.shape[1]
dimension
```

Inner Product Search Index (Works with Normalized vectors)
* Used for cosine similarity

```python
# Used to find vectors that have the highest dot product with a given query vector
# i.e. two vectors are most similar (they point in same direction and have greater magnitude)

text_index = faiss.IndexFlatIP(dimension)
image_index = faiss.IndexFlatIP(dimension)

# Add vectors to index
text_index.add(text_embeddings)
image_index.add(image_embeddings)

print("Text index size:", text_index.ntotal)
print("Image index size:", image_index.ntotal)
```

#### Save Indexes

```python
faiss.write_index(text_index, f"{PROJECT_DIR}/indexes/text_index.faiss")
faiss.write_index(image_index, f"{PROJECT_DIR}/indexes/image_index.faiss")
```
