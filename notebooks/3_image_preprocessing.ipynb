Import the packages

```python
import pandas as pd
from PIL import Image
import os
```

```python
PROJECT_DIR = '/content/drive/MyDrive/product_search_engine'
DATA_DIR = f"{PROJECT_DIR}/data"
```

```python
df = pd.read_csv(f"{DATA_DIR}/text_processed.csv")
```

```python
df.head()
```

```python
# List to store all the images that are present from the path mentioned in the dataframe
valid_images = []
```

```python
for path in df['images_path']:
  try:
    img = Image.open(path)
    img.verify() # Check if the image is truncated or broken
    valid_images.append(True)
  except:
    valid_images.append(False)
```

```python
df = df[valid_images]
```

```python
df.to_csv(f"{DATA_DIR}/image_validated.csv", index=False)
```
